```python
import undetected_chromedriver as uc
from bs4 import BeautifulSoup
import time
import os
import csv
import requests

def get_lamoda_data():
    # 1. Сначала подготавливаем папку и заголовки CSV
    if not os.path.exists('sneakers_img'):
        os.makedirs('sneakers_img')
    
    csv_file = 'sneakers.csv'
    file_exists = os.path.isfile(csv_file)
    
    # 2. Настраиваем браузер
    driver = uc.Chrome(version_main=146)
    
    try:
        url = "https://www.lamoda.ru/c/2968/shoes-krossovki-kedy/"
        driver.get(url)
        time.sleep(10) 
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        if "Доступ к сайту" not in soup.text:
            items = soup.find_all('div', class_='x-product-card__card')
            
            with open(csv_file, mode='a', newline='', encoding='utf-8') as f:
                writer = csv.writer(f, delimiter=';')
                if not file_exists:
                    writer.writerow(['brand', 'model', 'img_url', 'path_to_img'])

                for item in items: 
                    try:
                        brand = item.find('div', class_='x-product-card-description__brand-name').text.strip()
                        name = item.find('div', class_='x-product-card-description__product-name').text.strip()
                        img = item.find('img').get('src')
                        
                        # Улучшаем разрешение (убираем параметры обрезки)
                        raw_img = img.split('?')[0] 
                        full_img_url = "https:" + raw_img + "?width=1200&height=1800"
                        
                        file_name = f"{brand}_{name}".replace("/", "_").replace(" ", "_")[:50] + ".jpg"
                        file_path = os.path.join('sneakers_img', file_name)

                        # ТЕПЕРЬ writer доступен
                        writer.writerow([brand, name, full_img_url, file_path])

                        # Скачивание
                        img_data = requests.get(full_img_url, timeout=10).content
                        with open(file_path, 'wb') as img_file:
                            img_file.write(img_data)
                        
                        print(f"Успех: {brand} {name}")
                        time.sleep(0.2) # Маленькая пауза для стабильности
                    except Exception as e:
                        print(f"Ошибка на товаре: {e}")
                                
        else:
            print("Блокировка доступа.")
    finally:
        driver.quit()

get_lamoda_data()
```

```python
def get_lamoda_pages(pages_count=3):
    if not os.path.exists('sneakers_img'):
        os.makedirs('sneakers_img')
    
    csv_file = 'sneakers.csv'
    driver = uc.Chrome(version_main=146)
    
    try:
        with open(csv_file, mode='a', newline='', encoding='utf-8') as f:
            writer = csv.writer(f, delimiter=';')
            if os.stat(csv_file).st_size == 0:
                writer.writerow(['brand', 'model', 'img_url', 'path_to_img'])

            for page in range(1, pages_count + 1):
                print(f"--- Парсим страницу {page} ---")
                url = f"https://www.lamoda.ru/c/2968/shoes-krossovki-kedy/?page={page}"
                
                driver.get(url)
                time.sleep(7) # Ждем прогрузки страницы
                
                # Прокрутка вниз, чтобы подгрузились все Lazy-Load картинки
                driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
                time.sleep(3) 

                soup = BeautifulSoup(driver.page_source, 'html.parser')
                items = soup.find_all('div', class_='x-product-card__card')

                for item in items:
                    try:
                        brand = item.find('div', class_='x-product-card-description__brand-name').text.strip()
                        name = item.find('div', class_='x-product-card-description__product-name').text.strip()
                        
                        # Проверяем и src, и data-src для обхода Lazy Load
                        img_tag = item.find('img')
                        img = img_tag.get('data-src') or img_tag.get('src')
                        
                        if not img: continue

                        # Формируем HQ URL
                        clean_img = img.split('?')[0]
                        full_img_url = "https:" + clean_img + "?width=1200&height=1800"
                        
                        file_name = f"{brand}_{name}".replace("/", "_").replace(" ", "_")[:50] + ".jpg"
                        file_path = os.path.join('sneakers_img', file_name)

                        writer.writerow([brand, name, full_img_url, file_path])

                        # Скачивание (можно добавить проверку, если файл уже есть)
                        if not os.path.exists(file_path):
                            img_data = requests.get(full_img_url, timeout=10).content
                            with open(file_path, 'wb') as img_file:
                                img_file.write(img_data)
                        
                    except Exception as e:
                        print(f"Пропуск товара: {e}")
                
                print(f"Страница {page} готова.")

    finally:
        driver.quit()

get_lamoda_pages(100)
```

```python
import undetected_chromedriver as uc
from bs4 import BeautifulSoup
import time, os, csv, requests, json

def get_full_data():
    if not os.path.exists('lamoda_data'): os.makedirs('lamoda_data')
    driver = uc.Chrome(version_main=146)
    
    try:
        # 1. Получаем список ссылок на товары с первой страницы
        driver.get("https://www.lamoda.ru/c/2968/shoes-krossovki-kedy/")
        time.sleep(5)
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        links_tails = soup.find_all('a', class_="_root_aroml_2 _label_aroml_17 x-product-card__pic x-product-card__pic-catalog x-product-card__pic x-product-card__pic-catalog")
        links = ["https://www.lamoda.ru" + tail['href'] for tail in links_tails]

        with open('sneakers_full.csv', 'a', newline='', encoding='utf-8') as f:
            writer = csv.writer(f, delimiter=';')
            
            for url in links[:10]: # Для теста берем первые 10 товаров
                driver.get(url)
                time.sleep(3)
                
                # Ищем JSON с данными о товаре (там все фото)
                # Lamoda хранит данные в скрипте с id="__NEXT_DATA__" или в переменной __INITIAL_STATE__
                soup_item = BeautifulSoup(driver.page_source, 'html.parser')
                
                # Вытаскиваем бренд и название
                brand = soup_item.find('span', class_='x-premium-product-title__brand-name').text.strip()
                model = soup_item.find('div', class_='x-premium-product-title__model-name').text.strip()
                
                # СОБИРАЕМ РАКУРСЫ (из галереи)
                gallery_photos = []
                gallery_imgs = soup_item.find_all('img', class_='x-premium-product-gallery__image')
                for img in gallery_imgs:
                    src = img.get('data-src') or img.get('src')
                    if src: gallery_photos.append("https:" + src.split('?')[0] + "?width=1200")

                # СОБИРАЕМ ФОТО ИЗ ОТЗЫВОВ
                review_photos = []
                # Кликаем на вкладку отзывов, если нужно, или ищем в DOM
                review_imgs = soup_item.find_all('img', class_='re-image-gallery__image')
                for r_img in review_imgs:
                    r_src = r_img.get('src')
                    if r_src: review_photos.append(r_src)

                # Сохраняем в CSV
                writer.writerow([brand, model, "|".join(gallery_photos), "|".join(review_photos)])
                print(f"Собрано для {brand}: {len(gallery_photos)} ракурсов и {len(review_photos)} фото отзывов")

    finally:
        driver.quit()

get_full_data()
```

```python
from IPython.display import clear_output

def collect_lamoda_data(num_pages):   
    failed_pages = [] 
    for page in range(46, num_pages + 1):
        clear_output(wait=True)
        print(f"--- Парсим страницу {page} ---")
        url = f"https://www.lamoda.ru/c/2968/shoes-krossovki-kedy/?page={page}"
        driver = uc.Chrome(version_main=146)
        driver.get(url)
        time.sleep(7) # Ждем прогрузки страницы
        
        # Прокрутка вниз, чтобы подгрузились все Lazy-Load картинки
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(3) 

        soup = BeautifulSoup(driver.page_source, 'html.parser')
        #собираем все ссылки на карточки товаров
        try:
            links_tails = soup.find_all('a', class_="_root_aroml_2 _label_aroml_17 x-product-card__pic x-product-card__pic-catalog x-product-card__pic x-product-card__pic-catalog")
            links = ["https://www.lamoda.ru" + tail['href'] for tail in links_tails]
            for link in links:
                #заходим в карточку товара
                driver_item = uc.Chrome(version_main=146)
                driver_item.get(link)
                time.sleep(3)
                driver_item.execute_script("window.scrollTo(0, document.body.scrollHeight);")

                soup_item = BeautifulSoup(driver_item.page_source, 'html.parser')               
                
                #собираем бренд, модель, цвет, описание, категорию
                for script in soup_item.find_all('script', type="application/ld+json"):
                    try:
                        item_info = json.loads(script.string)[0]
                        brand = item_info["brand"]["name"]
                        model = item_info["name"]
                        images = item_info["image"]
                        category = item_info["category"]
                        description = item_info["description"]
                        print("Получены brand, model, images, category, description для ", brand, model)
                        #пробуем получить цвет
                        for script in soup_item.find_all('script', type="text/javascript"):
                            try:
                                match = re.search(r'customDimension19:"([^"]+)"', script.string)
                                if match:
                                    color = match.group(1)
                            except:
                                continue
                        #скачиваем все студийные фото 
                        i = 0
                        for img in images:
                            photo_url = "https:" + img
                            photo = requests.get(photo_url, timeout=10).content
                            file_name = f"{brand}_{model}".replace("/", "_").replace(" ", "_")[:50] + f"_{i}" + ".jpg"
                            file_path = os.path.join(lamoda_photo_path, file_name)
                            with open(file_path, 'wb') as f:
                                f.write(photo)
                                print(f"Скачано фото {i} для {brand} {model}")
                            #записываем в табличку собранную инфу по скачанной модели
                            with open(lamoda_csv_file, 'a', newline='', encoding='utf-8') as f:
                                writer = csv.writer(f, delimiter=';')
                                writer.writerow([brand, model, category, color, description, photo_url, file_path, images[0]])
                                print(f"Записана информация по {brand} {model}")
                            i += 1
                        #скачиваем все пользовательские фото 
                        try:
                            raw_review_data = soup_item.find_all("script", string=re.compile(r'var __NUXT__'))[0].string
                            json_part = raw_review_data.split('__NUXT__ =')[-1].strip()
                            if json_part.endswith(';'):
                                    json_part = json_part[:-1]
                            review_data = chompjs.parse_js_object(json_part)
                            photoreviews = review_data["payload"]["state"]["payload"]["product"]["photoreviews"]
                            i=0
                            for photoreview in photoreviews:
                                photo = requests.get(photoreview["url"], timeout=10).content
                                file_name = f"{brand}_{model}".replace("/", "_").replace(" ", "_")[:50] + f"_{i}" + ".jpg"
                                file_path = os.path.join(user_photo_path, file_name)
                                with open(file_path, 'wb') as f:
                                    f.write(photo)
                                with open(user_csv_file, 'a', newline='', encoding='utf-8') as f:
                                    writer = csv.writer(f, delimiter=';')
                                    path_to_title_photo = os.path.join(lamoda_photo_path, f"{brand}_{model}".replace("/", "_").replace(" ", "_")[:50] + "_0" + ".jpg")
                                    writer.writerow([brand, model, category, color, description, images[0], path_to_title_photo, photoreview["url"], file_path])
                                i += 1
                        except:
                            print("Не удалось получить пользовательские фото по модели ", model)
                            continue
                    except:
                        print("Не удалось получить данные по ", link)
                        
                    driver_item.quit()
                        
           
                driver_item.quit()

        except Exception as e:
            failed_pages.append(page)
            print(f"Не удалось получить ссылки на карточки товаров со страницы {page}. Ошибка {e}")

        finally:
            driver.quit()
```
